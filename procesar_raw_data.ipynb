```python
import pathlib
```

```python

INPUT_PATH = pathlib.Path("Stress-Predict-Dataset-main") / "Raw_data"

subjects = [f"S{i:02d}" for i in range(1, 36)]
subjects_folders = [INPUT_PATH / subject for subject in subjects]
tags_files = [subject_folder / f"tags_{subject}.csv" for subject_folder, subject in zip(subjects_folders, subjects)]
tags_files
```

```python
import csv
def read_tags_file(tags_file):
     with open(tags_file, "r") as f:
        reader = csv.reader(f)
        seconds = []
        for row in reader:
            seconds.append(int(row[0]))
     return seconds
```

```python
EDA_Path = subjects_folders[0] / "EDA.csv"

with open(EDA_Path, "r") as f:
    reader = csv.reader(f)
    eda_data = []
    for row in reader:
        eda_data.append(float(row[0]))

inicio = eda_data[0]
hz = eda_data[1]
seconds = read_tags_file(tags_files[0])
strop_test_start = seconds[0]

puntos_ignorados = int((strop_test_start - inicio) * hz)
puntos_ignorados
```

```python
strop_test_end = seconds[1]
puntos_label_1 = (strop_test_end - strop_test_start)*hz
label_list_1 = [1] * int(puntos_label_1)


relax_end = seconds[2]
puntos_label_2 = (relax_end - strop_test_end)*hz
label_list_2 = [0] * int(puntos_label_2)

question_end = seconds[3]
puntos_label_3 = (question_end - relax_end)*hz
label_list_3 = [1] * int(puntos_label_3)

interview_end = seconds[4]
puntos_label_4 = (interview_end - question_end)*hz
label_list_4 = [0] * int(puntos_label_4)

relax_end_2 = seconds[5]
puntos_label_5 = (relax_end_2 - interview_end)*hz
label_list_5 = [0] * int(puntos_label_5)

hyperventilation_end = seconds[6]
puntos_label_6 = (hyperventilation_end - relax_end_2)*hz
label_list_6 = [1] * int(puntos_label_6)

relax_end_3 = seconds[7]
puntos_label_7 = (relax_end_3 - hyperventilation_end)*hz
label_list_7 = [0] * int(puntos_label_7)

question_end_2 = seconds[8]
puntos_label_8 = (question_end_2 - relax_end_3)*hz
label_list_8 = [1] * int(puntos_label_8)

relax_end_4 = seconds[9]
puntos_label_9 = (relax_end_4 - question_end_2)*hz
label_list_9 = [0] * int(puntos_label_9)

complete_list = label_list_1 + label_list_2 + label_list_3 + label_list_4 + label_list_5 + label_list_6 + label_list_7 + label_list_8 + label_list_9


print(len(complete_list))
print(len(eda_data[2+puntos_ignorados:puntos_ignorados+len(complete_list)+2]))
```

```python
import numpy as np
from scipy import signal
def resample_1d(x, orig_fs, target_fs):
    x = np.asarray(x).squeeze()
    if orig_fs == target_fs:
        return x.astype(np.float32)
    duration = len(x) / float(orig_fs)
    target_len = max(2, int(round(duration * target_fs)))
    return signal.resample(x, target_len).astype(np.float32)
```
